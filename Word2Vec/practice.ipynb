```python
import gensim
import os
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
import spacy
```

```python
nlp = spacy.load("en_core_web_sm", disable=["parser",'ner'])
nlp.add_pipe("sentencizer")
nlp.max_length = 10000000
```

```python
from gensim.parsing.preprocessing import remove_stopwords

whole_story =[]
for filename in os.listdir(r"D:\nlp-basics\Word2Vec\harry potter books"):
    text = open(f"harry potter books\{filename}",'r',encoding='utf-8')
    doc = nlp(text.read())
    for sent in doc.sents:
        clean_text = remove_stopwords(sent.text)
        final_words_tokens = simple_preprocess(clean_text)
        whole_story.append(final_words_tokens)
```

```python
model = gensim.models.Word2Vec(window=5,min_count=2,sg=1)
model.build_vocab(whole_story)
model.train(whole_story,total_examples=model.corpus_count,epochs=30)
```

```python
try:
    print(model.wv.similarity('harry','ron'))
except KeyError as k:
    print(k)
```

```python
try:
    print(model.wv.most_similar('hogwarts'))
except KeyError as k:
    print(k)
```
